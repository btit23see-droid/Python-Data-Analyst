# 11. Operator Precedence

## Learning Objectives

- Understand operator precedence.
- Understand order of evaluation.
- Use parentheses.
- Compare expressions.

## Introduction

Operator precedence determines the order in which Python evaluates operators in an expression.

Calculate 10 + 5 * 2.

```python
result = 10 + 5 * 2
print(result)
```

Calculate (10 + 5) * 2.

```python
result = (10 + 5) * 2
print(result)
```



Compare the results of the previous two expressions.

```python
print(10 + 5 * 2)
print((10 + 5) * 2)
```



Calculate 20 / 5 + 2.

```python
print(20 / 5 + 2)
```

Calculate 20 / (9 + 3).

```python
print(20 / (9 + 3))
```



Create an expression using multiplication and addition.

```python
print(3 + 9 * 7)
```

Use parentheses to change the result.

```python
print((7 + 8) * 3)
```

Create an expression using subtraction and multiplication.

```python
print(4 - 7 * 2)
```

Use parentheses with division and addition.

```python
print(100 / (60 + 15))
```

Create your own expression containing at least three operators.

```python
result = (70 + 6) * 8 - 6
print(result)
```

## Observation

Multiplication and division are evaluated before addition and subtraction.
Parentheses can change the normal order of evaluation.

## Important Rules

- Parentheses have high priority.
- Multiplication and division are evaluated before addition and subtraction.
- Parentheses can change the result of an expression.

## Common Mistakes

- Ignoring operator precedence.
- Forgetting parentheses.
- Assuming every expression is evaluated simply from left to right.
