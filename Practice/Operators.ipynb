Add two numbers using the + operator.

```python
a = 10
b = 5

print(a - b)
```

```python
a = 10
b = 5

print(a * b)
```



Use division, modulus and floor division.

```python
a = 10
b = 3

print(a / b)
print(a % b)
print(a // b)
```



Use assignment operators to change a variable.

```python
x = 10

x += 5
print(x)

x -= 2
print(x)

x *= 2
print(x)
```



Compare two numbers using comparison operators.

```python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
```

```python
a = [1, 2]
b = a
c = [1, 2]

print(a is b)
print(a is c)
print(a is not c)
```

## Observation

Different operators are used for calculations, comparisons, assignments, logical conditions, identity checks and membership checks.

## Important Rules

- Arithmetic operators perform calculations.
- Assignment operators assign or update values.
- Comparison operators return True or False.
- Logical operators combine conditions.
- Identity operators use is and is not.
- Membership operators use in and not in.

## Common Mistakes

- Confusing = and ==.
- Confusing / and //.
- Confusing is with ==.
- Forgetting that comparison results are Boolean values.
