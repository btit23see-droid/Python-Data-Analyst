Print numbers from 1 to 10


A for loop is used to repeat a block of code. Here, it prints numbers from 1 to 10.

```python
for i in range(1, 11):
    print(i)
```

Print even numbers from 1 to 20


This program prints only even numbers from 1 to 20. We use a step of 2 to skip odd numbers.

```python
for i in range(2, 21, 2):
    print(i)
```

Print odd numbers from 1 to 20


This program prints only odd numbers from 1 to 20. It starts from 1 and increases by 2.

```python
for i in range(1, 21, 2):
    print(i)
```

Print all elements of a list

A for loop can be used to access and print each element of a list one by one.

```python
fruits = ["orange", "papaya", "kiwi"]

for fruit in fruits:
    print(fruit)
```

Print each character of a string


A string contains characters. The for loop prints each character one by one.

```python
name = "Saisa"

for letter in name:
    print(letter)
```

Print multiplication table of 5


This program prints the multiplication table of 5 from 1 to 10.

```python
for i in range(1, 11):
    print("5 x", i, "=", 5 * i)
```

Print squares of numbers from 1 to 10

Explanation:
This program finds the square of each number from 1 to 10. A square is calculated by multiplying a number by itself.

```python
for i in range(1, 11):
    print(i * i)
```

Print numbers in reverse order (10 to 1)


This program prints numbers from 10 down to 1. -1 means the number decreases by 1 each time.

```python
for i in range(10, 0, -1):
    print(i)
```

Print all vowels in a string


This program checks each character in a string and prints it if it is a vowel (a, e, i, o, u).

```python
name = "Sushant"

for letter in name:
    if letter in "aeiou":
        print(letter)
```

Print sum of numbers from 1 to 10


This program adds all numbers from 1 to 10 and stores the result in the total variable.

```python
total = 0

for i in range(1, 11):
    total = total + i

print(total)
```
