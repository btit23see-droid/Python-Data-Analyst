Print numbers from 1 to 10


A while loop repeats the code as long as the given condition is True.

```python
i = 1

while i <= 10:
    print(i)
    i = i + 1
```

Print numbers from 10 to 1


This program prints numbers from 10 to 1 in reverse order. The value decreases by 1 each time.

```python
i = 10

while i >= 1:
    print(i)
    i = i - 1
```

Print even numbers up to 50

This program prints even numbers from 2 to 50. The value increases by 2 each time.

```python
i = 2

while i <= 50:
    print(i)
    i = i + 2
```

Print sum of first 10 numbers

This program adds the first 10 numbers and stores the total in the total variable.

```python
i = 1
total = 0

while i <= 10:
    total = total + i
    i = i + 1

print(total)
```

Print multiplication table of 3


This program prints the multiplication table of 3 from 1 to 10.

```python
i = 1

while i <= 30:
    print("3 x", i, "=", 3 * i)
    i = i + 1
```

Count digits in a number

This program counts how many digits are present in a number. // 10 removes the last digit each time.

```python
num = 12345
count = 0

while num > 0:
    num = num // 10
    count = count + 1

print(count)
```

Reverse a number

This program reverses a number. % 10 gets the last digit and // 10 removes the last digit.

```python
num = 1234
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print(reverse)
```

Print squares of numbers from 1 to 10

This program prints the square of every number from 1 to 10.

```python
i = 1

while i <= 10:
    print(i * i)
    i = i + 1
```

Print all digits of a number

This program extracts and prints each digit of a number. % 10 is used to get the last digit.

```python
num = 12345

while num > 0:
    digit = num % 10
    print(digit)
    num = num // 10
```

Print factorial of a number

Factorial means multiplying a number by all positive numbers below it. For example, 5! = 5 × 4 × 3 × 2 × 1 = 120.

```python
num = 7
fact = 1
i = 1

while i <= num:
    fact = fact * i
    i = i + 1

print(fact)
```
