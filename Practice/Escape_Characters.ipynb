# 08. Escape Characters

## Learning Objectives

- Understand escape characters.
- Use new lines.
- Use tabs.
- Display quotation marks.
- Display backslashes.

## Introduction

Escape characters are special characters used inside strings to create special formatting.

## Common Escape Characters

\n = new line
\t = tab
\" = double quote
\' = single quote
\\ = backslash



Print Hello and World on separate lines.

```python
print("Hello\nWorld")
```



Print your name and age using a tab.

```python
print("Name:\tSeema")
print("Age:\t22")
```



Print a sentence containing double quotation marks.

```python
print("She said \"Hello\"")
```



Print a sentence containing a single quotation mark.

```python
print('It\'s Python')
```



Print a Windows-style file path.

```python
print("C:\\Users\\Phabish")
```



Print three words on three separate lines.

```python
print("Python\nJava\nC++")
```



Display student information using tabs.

```python
print("Name:\tSushant")
print("Age:\t26")
print("Department:\tIT")
```



Use both newline and tab characters.

```python
print("Student\n\tName: Saisa\n\tAge: 2")
```



Print an address using new lines.

```python
print("Saisa Majhi\nKathmandu\nNepal")
```



Create a formatted student record using escape characters.

```python
print("Student Information\n\tName: Seema\n\tAge: 22\n\tDepartment: IT")
```

# Observation

Escape characters change the formatting and display of strings.

## Important Rules

- \n creates a new line.
- \t creates a tab.
- \" displays a double quote.
- \' displays a single quote.
- \\ displays a backslash.

## Common Mistakes

- Using escape characters incorrectly.
- Forgetting the backslash.
- Confusing \n with normal text.
