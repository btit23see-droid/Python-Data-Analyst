## Learning Objectives

- Understand lists.
- Create lists.
- Access list elements.
- Use positive and negative indexing.
- Slice lists.
- Change list elements.
- Add and remove elements.
- Sort and reverse lists.
- Search and count items.
- Clear a list.

## Introduction

A list is a collection used to store multiple values in one variable.

Lists are ordered and changeable.



Create a list containing five student names.

```python
students = ["Seema", "Ram", "Sita", "Hari", "John"]

print(students)
```


Print the first and last student.

```python
students = ["Seema", "Ram", "Sita", "Hari", "John"]

print(students[0])
print(students[-1])
```



Print the first three students using slicing.

```python
students = ["Seema", "Ram", "Sita", "Hari", "John"]

print(students[0:3])
```



Change the second student's name.

```python
students = ["Seema", "Ram", "Sita"]

students[1] = "John"

print(students)
```



Add a new student using append().

```python
students = ["Seema", "Ram", "Sita"]

students.append("Hari")

print(students)
```



Remove a student from the list.

```python
students = ["Seema", "Ram", "Sita"]

students.remove("Sita")

print(students)
```



Find the number of students using len().

```python
students = ["Seema", "Ram", "Sita"]

print(len(students))
```



Sort the list alphabetically.

```python
students = ["Seema", "Ram", "Saisa", "Ayanshi"]

students.sort()

print(students)
```



Search for a student and count how many times the name appears.

```python
students = ["Seema", "sushant", "Saisa", "Phabish"]

print("Seema" in students)
print(students.count("Seema"))
```



Reverse and then clear a list.

```python
students = ["Seema", "Ram", "Sita"]

students.reverse()
print(students)

students.clear()
print(students)
```

## Observation

Lists can contain multiple values and can be changed after creation.
List methods are used to add, remove, search, sort, reverse and clear elements.

## Important Rules

- List indexing starts from 0.
- Negative indexing starts from the end.
- Lists are changeable.
- append() adds an item.
- remove() removes an item.
- len() returns the number of items.
- sort() sorts items.
- reverse() reverses items.
- clear() removes all items.

## Common Mistakes

- Using an invalid index.
- Forgetting that indexing starts from 0.
- Removing an item that does not exist.
- Using the wrong list method.
