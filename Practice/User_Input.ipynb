# 05. User Input

## Learning Objectives

- Understand input().
- Take information from users.
- Store user input.
- Convert input into numbers.
- Use input in calculations.

## Introduction

The input() function allows a program to receive information from the user.

By default, input() returns a string.

## Detailed Explanation

Syntax:

variable = input("message")



Ask the user to enter their name and print it.

```python
name = input("Enter your name: ")
print("Hello", name)
```



Ask the user to enter their age and print it.

```python
age = int(input("Enter your age: "))
print("Your age is", age)
```



Take two numbers from the user and calculate their sum.

```python
num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

print("Sum:", num1 + num2)
```



Take two numbers and calculate their difference.

```python
num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

print("Difference:", num1 - num2)
```



Take two numbers and calculate their multiplication.

```python
um1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

print("Multiplication:", num1 * num2)
```



Take two numbers and calculate their division.

```python
num1 = float(input("Enter number 1: "))
num2 = float(input("Enter number 2: "))

print("Division:", num1 / num2)
```



Take three numbers and calculate their average.

```python
num1 = float(input("Enter number 1: "))
num2 = float(input("Enter number 2: "))
num3 = float(input("Enter number 3: "))

average = (num1 + num2 + num3) / 3

print("Average:", average)
```



Take length and breadth from the user and calculate the area of a rectangle.

```python
length = float(input("Enter length: "))
breadth = float(input("Enter breadth: "))

area = length * breadth

print("Area:", area)
```



Take temperature in Celsius and convert it to Fahrenheit.

```python
celsius = float(input("Enter Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print("Fahrenheit:", fahrenheit)
```



Take student name and marks from the user and display them.

```python
name = input("Enter student name: ")
marks = float(input("Enter marks: "))

print("Student:", name)
print("Marks:", marks)
```

## Observation

User input allows programs to interact with users.
Numeric input normally needs conversion before mathematical operations.

## Important Rules

- input() normally returns a string.
- Use int() for whole numbers.
- Use float() for decimal numbers.
- Convert input before performing numerical calculations.

## Common Mistakes

- Forgetting int() or float().
- Entering text when a number is required.
- Performing mathematical operations directly on string input.
