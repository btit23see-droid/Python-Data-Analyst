# 01. Variables

## Learning Objectives

- Understand variables.
- Learn how to create and assign variables.
- Understand variable naming rules.
- Learn valid and invalid variable names.
- Understand case sensitivity.
- Learn how to change variable values.
- Check the type of a variable.

## Introduction

A variable is a name used to store a value in Python.

Variables allow us to store information and use it later in a program.

## Detailed Explanation

A variable is created using the assignment operator (=).

Syntax:

variable_name = value


Create a variable called name and store your name in it.

```python
name = "Seema"
print(name)
```



Create a variable called age and store your age.

```python
age = 22
print(age)
```



Create a variable called department and store your department.

```python
department = "IT"
print(department)
```



Create a variable called college and store your college name.

```python
college = "ISMT College"
print(college)
```



Create a variable containing a decimal number.

```python
height = 5.4
print(height)
```



Create a Boolean variable called student.

```python
student = True
print(student)
```


Create three variables and print all of them.

```python
name = "Seema"
age = 22
department = "IT"

print(name)
print(age)
print(department)
```



Change the value of age from 22 to 23.

```python
age = 22
age = 23

print(age)
```



Check the data type of the age variable.

```python
age = 22

print(type(age))
```



Create three variables in one line and print them.

```python
name, age, city = "Seema", 22, "Kathmandu"

print(name)
print(age)
print(city)
```

## Observation

Variables store values that can be accessed and changed during program execution.
Python automatically determines the type of the stored value.

## Important Rules

- Variable names cannot start with a number.
- Variable names cannot contain spaces.
- Variable names are case-sensitive.
- Use meaningful variable names.
- Python keywords cannot be used as variable names.

## Common Mistakes

- Starting a variable name with a number.
- Using spaces in variable names.
- Misspelling variable names.
- Confusing uppercase and lowercase letters.
