# 04. Data Types

## Learning Objectives

- Understand Python data types.
- Identify int, float, str and bool.
- Understand list, tuple, set and dictionary.
- Understand None.
- Check data types using type().

## Introduction

A data type tells Python what kind of value is stored in a variable.

## Detailed Explanation

Common Python data types include:

int - whole numbers
float - decimal numbers
str - text
bool - True or False
list - collection of values
tuple - ordered collection
set - unique values
dict - key-value pairs
None - no value



Create an integer variable and check its type.

```python
age = 22
print(type(age))
```



Create a float variable and check its type.

```python
height = 5.4
print(type(height))
```



Create a string variable and check its type.

```python
name = "Seema"
print(type(name))
```



Create a Boolean variable and check its type.

```python
student = True
print(type(student))
```



Create a list and check its type.

```python
students = ["Seema", "Ram", "Sita"]
print(type(students))
```



Create a tuple and check its type.

```python
marks = (80, 90, 75)
print(type(marks))
```



Create a set and check its type.

```python
subjects = {"Python", "Database", "Web"}
print(type(subjects))
```



Create a dictionary and check its type.

```python
student = {"name": "Seema", "age": 22}
print(type(student))
```



Create a variable with None and check its type.

```python
result = None
print(type(result))
```



Create four variables with different data types and print their types.

```python
name = "Seema"
age = 22
height = 5.4
student = True

print(type(name))
print(type(age))
print(type(height))
print(type(student))
```

## Observation

Python provides different data types for different kinds of values.
The type() function can identify the type of a value.

## Important Rules

- int stores whole numbers.
- float stores decimal values.
- str stores text.
- bool stores True or False.
- Collections can store multiple values.

## Common Mistakes

- Forgetting quotation marks for strings.
- Confusing 10 with "10".
- Using True and False as strings instead of Boolean values.
