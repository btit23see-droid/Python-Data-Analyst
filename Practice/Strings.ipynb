## Learning Objectives

- Understand strings.
- Create strings.
- Find string length.
- Use indexing.
- Use negative indexing.
- Slice strings.
- Use string methods.

## Introduction

A string is a sequence of characters used to store text.

## Detailed Explanation

Strings can be created using single or double quotation marks.
Indexing starts from 0.



Create a string containing your name.

```python
name = "Seema"
print(name)
```



Find the length of your name.

```python
name = "Seema"
print(len(name))
```



Print the first character of your name.

```python
name = "Seema"
print(name[0])
```



Print the last character using negative indexing.

```python
name = "Seema"
print(name[-1])
```



Print the first three characters of your name.

```python
name = "Phabish"
print(name[0:3])
```



Convert a string to uppercase.

```python
text = "hello python"
print(text.upper())
```



Convert a string to lowercase.

```python
text = "HELLO PYTHON"
print(text.lower())
```



Convert a sentence to title case.

```python
text = "python is easy to learn"
print(text.title())
```



Replace the word "Python" with "Java".

```python
text = "I am learning Python"
print(text.replace("Python", "Java"))
```



Print the first and last characters of a string.

```python
text = "Programming"

print(text[0])
print(text[-1])
```

## Observation

Indexing accesses individual characters, while slicing accesses a part of a string.
String methods can change how text is displayed or processed.

## Important Rules

- String indexing starts from 0.
- Negative indexing starts from the end.
- len() returns the number of characters.
- Strings are immutable.

## Common Mistakes

- Using an index that does not exist.
- Forgetting that indexing starts from 0.
- Trying to directly change one character of a string.
