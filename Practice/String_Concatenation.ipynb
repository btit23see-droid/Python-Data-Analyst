# 07. String Concatenation

## Learning Objectives

- Understand string concatenation.
- Join strings using +.
- Combine variables and strings.
- Use str() for conversion.
- Use f-strings.

## Introduction

String concatenation means joining two or more strings together.
The + operator can be used to join strings.



Join your first name and last name.

```python
first_name = "Sushant"
last_name = "Chaudhary"

full_name = first_name + " " + last_name

print(full_name)
```



Join two city names.

```python
city1 = "Kathmandu"
city2 = "Pokhara"

print(city1 + " and " + city2)
```



Create a sentence using two string variables.

```python
name = "Seema"
department = "IT"

print("My name is " + name + " and I study " + department)
```



Join a string and an integer using str().

```python
name = "Seema"
age = 22

print("My age is " + str(age))
```



Create a sentence using an f-string.

```python
name = "Seema"
age = 22

print(f"My name is {name} and I am {age} years old.")
```



Display your name and department using an f-string.

```python
name = "Seema"
department = "IT"

print(f"My name is {name} and my department is {department}.")
```



Combine three strings.

```python
a = "Python"
b = "is"
c = "easy"

print(a + " " + b + " " + c)
```



Create a student introduction using concatenation.

```python
name = "Seema"
age = 22
college = "ISMT"

print("My name is " + name + ", I am " + str(age) + " years old and I study at " + college + ".")
```



Combine a product name and price using an f-string.

```python
product = "Laptop"
price = 80000

print(f"The price of {product} is Rs. {price}.")
```



Create a complete sentence using three variables.

```python
name = "Seema"
age = 22
city = "Kathmandu"

print(f"My name is {name}, I am {age} years old and I live in {city}.")
```

## Observation

Strings can be joined using +.
Numbers need conversion when concatenated using +, while f-strings can directly include variables.

## Important Rules

- + can concatenate strings.
- Use str() when joining a number with a string using +.
- f-strings provide an easy way to combine text and variables.

## Common Mistakes

- Concatenating a string and integer without conversion.
- Forgetting spaces between strings.
- Forgetting quotation marks.
