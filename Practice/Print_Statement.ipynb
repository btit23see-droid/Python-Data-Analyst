# 02. Print Statement

## Learning Objectives

- Understand print().
- Print text and numbers.
- Print variables.
- Print multiple values.
- Understand sep and end.


## Introduction

The print() function is used to display output on the screen.

## Detailed Explanation

The basic syntax is:

print(value)

The value can be text, numbers, variables, or expressions.



Print "Hello World".

```python
print("Hello World")
```



Print your name.

```python
print("Seema")
```



Print your age.

```python
print(22)
```



Print the result of 10 + 20.

```python
print(10 + 20)
```



Create a name variable and print it.

```python
name = "Seema"
print(name)
```



Print your name and age together.

```python
name = "Seema"
age = 22

print("Name:", name)
print("Age:", age)
```



Print three words using sep="-".

```python
print("Python", "is", "easy", sep="-")
```



Use end=" " to print two words on the same line.

```python
print("Hello", end=" ")
print("Seema")
```



Print your name, age and department in one print statement.

```python
name = "Seema"
age = 22
department = "IT"

print(name, age, department)
```



Print a complete sentence about yourself.

```python
name = "Seema"
age = 22

print("My name is", name, "and I am", age, "years old.")
```

## Observation

The print() function displays values as output.
The sep and end arguments can control how the output is displayed.

## Important Rules

- Use parentheses with print().
- Text should normally be inside quotation marks.
- Variables do not need quotation marks.
- sep controls the separator.
- end controls the ending.

## Common Mistakes

- Forgetting parentheses.
- Forgetting quotation marks around text.
- Putting a variable inside quotes when its value is required.
