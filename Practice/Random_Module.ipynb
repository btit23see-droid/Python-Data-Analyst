## Introduction

The random module provides functions for generating random values and making random selections.

## Detailed Explanation

The random module can be imported using:

import random

Common functions include:

randint()
random()
choice()
sample()
shuffle()


Generate a random integer between 1 and 10.

```python
import random

number = random.randint(1, 10)

print(number)
```



Generate a random integer between 1 and 100.

```python
import random

number = random.randint(1, 100)

print(number)
```


Generate a random decimal value.

```python
import random

number = random.random()

print(number)
```



Select one random student from a list.

```python
import random

students = ["Seema", "Ram", "Sita", "Hari"]

print(random.choice(students))
```



Select two random students from a list.

```python
import random

students = ["cutie", "sansa", "Sweaty", "fire"]

print(random.sample(students, 2))
```



Shuffle a list of students.

```python
import random

students = ["Saina", "Rashmika", "Sunita", "Anita"]

random.shuffle(students)

print(students)
```



Create a simple dice simulator.

```python
import random

dice = random.randint(1, 6)

print("Dice:", dice)
```



Create a coin toss program.

```python
import random

result = random.choice(["Heads", "Tails"])

print(result)
```



Create a simple number guessing game.

```python
import random

number = random.randint(3, 50)

guess = int(input("Guess a number between 3 and 50: "))

if guess == number:
    print("Correct!")
else:
    print("Wrong! The number was", number)
```



Create a simple Love Calculator using a random percentage.

```python
import random

boy = input("Enter boy name: ")
girl = input("Enter girl name: ")

love = random.randint(1, 100)

print(boy, girl)
print("Love percentage:", love, "%")

if love > 90:
    print("Perfect Match")
elif love >= 50:
    print("Good Match")
else:
    print("Just Friends")
```

## Observation

The random module produces different random results when the program is run.
It can be used for games, simulations and random selections.

## Important Rules

- Import random before using its functions.
- randint() generates a random integer.
- random() generates a random decimal.
- choice() selects one item.
- sample() selects multiple items.
- shuffle() changes the order of items.

## Common Mistakes

- Forgetting to import random.
- Using incorrect function names.
- Giving an invalid range.
- Trying to select more items than available using sample().
