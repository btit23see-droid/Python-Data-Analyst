# 09. Type Casting

## Learning Objectives

- Understand type casting.
- Convert values between data types.
- Use int(), float(), str() and bool().
- Convert user input.
- Understand conversion errors.

## Introduction

Type casting means converting a value from one data type to another.

## Detailed Explanation

Common conversion functions are:

int()
float()
str()
bool()



Convert the string "22" into an integer.

```python
age = "14"

age = int(age)

print(age)
print(type(age))
```



Convert an integer into a float.

```python
number = 20

print(float(number))
```



Convert an integer into a string.

```python
number = 30

number = str(number)

print(number)
print(type(number))
```



Convert an integer into a Boolean value.

```python
number = 10

print(bool(number))
```



Take a number as input and convert it to integer.

```python
number = int(input("Enter a number: "))

print(number)
```



Take a decimal number as input.

```python
number = float(input("Enter a decimal number: "))

print(number)
```



Take two numbers and add them after conversion.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(a + b)
```



Convert the string "100" to an integer and multiply it by 2.

```python
number = int("100")

print(number * 2)
```



Convert a number to a string and join it with text.

```python
age = 53

print("Age: " + str(age))
```



Convert "5.5" into a float and add 2.

```python
number = float("5.2")

print(number + 2)
```

## Observation

Type casting allows values to be converted into different data types.
It is especially useful when working with user input.

## Important Rules

- int() converts suitable values into integers.
- float() converts suitable values into decimal numbers.
- str() converts values into strings.
- bool() converts values into Boolean values.
- Invalid conversions can cause errors.

## Common Mistakes

- Trying to convert invalid text into an integer.
- Forgetting to convert input before calculation.
- Confusing "10" with 10.
