# 03. Comments

## Learning Objectives

- Understand comments.
- Use single-line comments.
- Understand multi-line explanations.
- Understand why comments are useful.

## Introduction

Comments are notes written in a Python program to explain the code.
Python does not execute normal comments.

## Detailed Explanation

A single-line comment starts with the # symbol.


Write a comment explaining that the next line stores a name.

```python
# Store the student's name
name = "Seema"
print(name)
```



Write a comment before printing your age.

```python
age = 22

# Print the age
print(age)
```



Comment out a print statement.

```python
# print("This line will not execute")

print("This line will execute")
```



Add a comment explaining an addition operation.

```python
# Add two numbers
a = 10
b = 20

print(a + b)
```



Add comments to a subtraction program.

```python
# Store two numbers
a = 20
b = 5

# Subtract b from a
print(a - b)
```



Add comments to a multiplication program.

```python
# Multiply two numbers
a = 5
b = 4

print(a * b)
```



Add comments to an input program.

```python
# Ask the user for their name
name = input("Enter your name: ")

print(name)
```



Write a multi-line explanation using triple quotes.

```python
"""
This program stores
and prints a student's name.
"""

name = "Seema"
print(name)
```



Add comments to a program that calculates an average.

```python
# Store three numbers
a = 10
b = 20
c = 30

# Calculate average
average = (a + b + c) / 3

print(average)
```



Write comments for a simple student information program.

```python
# Student information
name = "Seema"
age = 22
department = "IT"

# Display student information
print(name)
print(age)
print(department)
```

## Observation

Comments make programs easier to understand and explain.

## Important Rules

- Use # for single-line comments.
- Comments should explain useful information.
- Comments are not normally executed as program instructions.

## Common Mistakes

- Forgetting # for a single-line comment.
- Writing unnecessary comments.
- Confusing comments with actual program code.
